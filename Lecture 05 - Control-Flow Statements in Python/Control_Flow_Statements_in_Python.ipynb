# **Control-Flow Statements in Python**

- for loop
- while loop



- steps to perform a loop -
  - variable declaration (Starting Point)
  - Applying Condition (to provide an end point)
  - incre | decre


```python
num = 1

while (num <= 10):
  print(num)
  num += 1
```

```python
# print only even numbers from 1 to 100
import time as tl
start_time = tl.time()
number = 1

while(number <= 100):
  if number%2 == 0:
    print(number)
  number += 1
end_time = tl.time()

print(end_time - start_time)
```

```python
import time as tl

startTime = tl.time()
num = 2

while(num <= 100):
  print(num)
  num += 2
endTime = tl.time()

print(endTime - startTime)
```

```python
import time

start_time = time.time()
for num in range(1, 101, 1):
  if(num%2 == 0):
    print(num)
end_time = time.time()

print(end_time - start_time)
```
