# **Exception Handling in Python**

- In Exception Handling we always use *Try*  and  *Except*  block
- Try block will always be used in the operation body
- Except block will always be used in the error handling

### **Handling TypeError Execption**

```python
try:
  num = 12
  string = "JavaScript"

  print(num + string)

except TypeError:
  print("You are trying to add two different data-type elements! \nIn python you can only add or operate similar type of data elements!")
```

### **Handling ZeroDivision Error**

```python
try:
  num = 0
  value = 68

  print(value/num)

except ZeroDivisionError:
  print("you're trying to divide a value with zero which directly results in to non-finite values which an exception in python. \nPlease try not to divide anything with zero!")
```
