# **Functions in Python**

- it is a bunch of code or a bundle of code or a container of code which will perform certain task
- functions are re-usable (we can call a function again and again without using lengthy codes)
- to declare a functions in python we need to declare it with the 'def' keyword
  - Syntax -
    def functionName():
      function body
- to execute a function in python we need to call it, coz without calling a function we can't execute any functionality

- Types of function -
  - User-Defined Function
  - Pre-Defined Function (built-in or pre-built function)

### - In User-Defined Function there are three classifications (these classifications are not official, classified by Bhavesh sir only)

User-Defined Functions -
- Standard function (non-parametric)
- Parametric Function
- Function with retured value

```python
# Standard Function -

def alertMsg():  # declaring a function
  print("This is alert message!")


alertMsg()       # calling a function
```

```python
# Parametric Function -

def sum(n1, n2=21):     # while declaring a function the key or value used in the functional brace '()' will be known as Parameters
  num1 = n1
  num2 = n2

  result = num1 + num2
  print(result)

sum(41,25)  # whenever we'll call a function, the value passed while calling a function will be known as Arguments
sum(56,86)
sum(12,32)
sum(42)
sum(96,20)

```

```python
# Function with Return Values
total_marks = 500
student_name = "Sarthak Sharma"

def obtained_marks(en, hn, maths, sc, sst):
  ob_marks = en + hn + maths + sc + sst
  return ob_marks


def percentage(tt_marks, ob_marks):
  percentage = (ob_marks * 100) / tt_marks
  return percentage


def grading_system(std_name, ob_marks, tt_marks, percentage):
  if percentage >= 80 and percentage <= 100:
    # print(std_name, "has scored", ob_marks, "out of", tt_marks, "and secured A Grade")
    print(f"{std_name} has scored {ob_marks} out of {tt_marks} and secred A Grade")
  elif percentage >= 60 and percentage < 80:
    print(f"{std_name} has scored {ob_marks} out of {tt_marks} and secred B Grade")
  elif percentage >= 40 and percentage < 60:
    print(f"{std_name} has scored {ob_marks} out of {tt_marks} and secred C Grade")
  elif percentage >= 33 and percentage < 40:
    print(f"{std_name} has scored {ob_marks} out of {tt_marks} and secred D Grade")
  else:
    print(f"{std_name} has failed!")



obtain_marks = obtained_marks(90,90,90,90,90)
print(obtain_marks)
percent_value = percentage(total_marks, obtain_marks)
print(percent_value)

grading_system(student_name, obtain_marks, total_marks, percent_value)
```
