# **Data-Types in Python**

- Type of Data stored in a variable is known as Data-Type
- Data-types are of two types
  - Primitive Data-Type - int | float | str | boolean
  - Non-Primitive Data-Type (Reference Data-Type) - list | set | tuple | dict



### **Primitive Data-Type -**

```python
# int data-type
taskCounter = 0
print(type(taskCounter))


taskCounter += 13
print(type(taskCounter))

# Float
taskCounter += 13.5
print(type(taskCounter))

# String
taskCounter1 = '13.5'
taskCounter = "13.5"
print(type(taskCounter + taskCounter1))
print(taskCounter + taskCounter1)

# Bool
flag = True
print(flag)
print(type(flag))
```

### **Set In Python**
- is a no-structure collection of the similar and the different data type elements stored or wrapped between '{}'
- Set doesn't have any index values
- therefore we can't perform indexing (we'll not able to target a specific value from the set)

```python
set1 = {10, 50, 50,20,30,40,50,20,60}
print(set1)
print(type(set1))

# set1[0]

```

### **Dictionary in Python**

- it is also a well-defined collection of data elements but in key:value pair
- in dict there is no integer index values are applicable, instead of int indexes we'll use keys as a index value
- denoted by '{}'

```python
empDi = {
    "empName": "Mahak Pandit",
    "empGender": "Female",
    "empSalary": 24,
    "empMail": "mahak@gmail.com"
    }
# print(empDi)
# print(type(empDi))

print(empDi["empMail"])
print(empDi['empName'])
```

# **Operators in Python**

- operation - is a process through which we can perform some analytical or mathematical task
- operators - are the symbols which will be responsible to perform the operations
- operands  - these are the values on which we perform operations


Types of Operators in Python -
1. Arithmetic Operators :- +, -, /, *, //, **, %
2. Assignment Operators :- +=, -=, /=, *=, //=, **=, %=
3. Comparison Operators (Conditional Operators) :- ==, !=, >, >=, <, <=
4. Logical Operators :- AND | OR | NOT'
5. Membership Operators :- in | not in
6. Identity Operators :- is | is not
7. BitWise Operators :- AND '&' | OR '|' | XOR '^'
