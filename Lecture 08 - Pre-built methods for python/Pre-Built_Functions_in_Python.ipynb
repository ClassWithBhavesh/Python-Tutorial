# **Pre-Built Functions (Built-In Functions)**


### **Built-In functions for LIST**

```python
# 1. li.sort() -

bike_speed = [45, 55, 75, 105, 68, 102]
print(bike_speed)

bike_speed.sort()
print(bike_speed)

bike_speed.sort(reverse=True)
print(bike_speed)
```

```python
# 2. li.append() -

bike_speed = [45, 55, 75, 105, 68, 102]
print(bike_speed)

bike_speed.append(112)
print(bike_speed)
```

```python
# 3. li.extend() -
li1 = [10,20,30]
li2 = [40,50,60]
print(li1)
print(li2)

li = li1 + li2
print(li)

li1.pop()
print(li1)
print(li2)
print(li)

# here comes the extend -
li1.extend(li2)
print(li1)
print(li2)
```

```python
# 4. li.insert(value)

bike_speed = [45, 55, 75, 105, 68, 102]
print(bike_speed)

bike_speed.insert(len(bike_speed) - 1, 112)
print(bike_speed)

```

```python
# 5. li.remove(value) -

bike_speed = [45, 55, 75, 105, 75, 68, 102, 75]
print(bike_speed)

bike_speed.remove(75)
print(bike_speed)
```

```python
# 6. li.pop() -
bike_speed = [45, 55, 75, 105, 75, 68, 102, 75]
print(bike_speed)

bike_speed.pop(-2)
bike_speed.pop(2)
bike_speed.pop(4)
print(bike_speed)
```

```python
# 7. li.clear() -

bike_speed = [45, 55, 75, 105, 75, 68, 102, 75]
print(bike_speed)

bike_speed.clear()
print(bike_speed)
```

```python
# 8. index(value, start, end) -

bike_speed = [45, 55, 75, 105, 75, 68, 102, 75]
print(bike_speed)

ind_value = bike_speed.index(75)
print(ind_value)

ind_value = bike_speed.index(75, -5, -3)
print(ind_value)
```

```python
# .9 li.count() -

bike_speed = [45, 55, 75, 105, 75, 68, 102, 75]
print(bike_speed)

print(bike_speed.count(75))
```

```python
# 10. reverse() -
bike_speed = [45, 55, 75, 105, 75, 68, 102, 75]
print(bike_speed)

bike_speed.reverse()
print(bike_speed)
```

```python
# 11. li.copy() -

bike_speed = [45, 55, 75, 105, 75, 68, 102, 75]
print(bike_speed)

new_bike_speed = bike_speed.copy()
print(new_bike_speed)


bike_speed.pop()
print(bike_speed)
print(new_bike_speed)
```
