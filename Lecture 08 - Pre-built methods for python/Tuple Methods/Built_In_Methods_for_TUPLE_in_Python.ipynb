# **Pre-Built Functions (Built-In Functions)**


### **Built-In functions for TUPLE**

```python
# 1. tup.count()

tup = (1, 4, 6,2, 4,3,6,6,4,5, 6, 4,6)
print(tup)

count_of_4 = tup.count(6)
print(count_of_4)
```

```python
# 2. tup.index()

tup = ("Aman", "Sakshi", "Himanshi", "Himanshu", "Khushi", "Rahul")
print(tup)

index_of_Sakshi = tup.index("Himanshi")
print(index_of_Sakshi)
```
