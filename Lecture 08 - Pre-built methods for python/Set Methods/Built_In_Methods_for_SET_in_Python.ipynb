# **Pre-Built Functions (Built-In Functions)**


### **Built-In functions for SET**

```python
# 1. set.add(value) :
set1 = {10,20,50,30,60,90}
print(set1)

set2 = "Bhagyashree"
print(set2)

set1.add(set2)
set1.add(True)
print(set1)
print(set2)
```

```python
# 2. set.clear()
set1.clear()

print(set1)
```

```python
# 3. set.copy()

setA = {'A', 'B', 'b', 'C', 'a', 'c'}
# print(setA)

setB = setA.copy()
print(setA)
print(setB)

setB.add("Z")
print(setA)
print(setB)

```

```python
# 4. set.difference(anotherSet)
setA = {'A', 'B', 'b', 'C', 'a', 'c'}
setB = {'a', 'B', 'B', 'C', 'A', 'C'}

print(setA)
print(setB)
print(setA.difference(setB))
print(setA)
print(setB)
```

```python
# 5. set.difference_update(anotherSet)
setA = {'A', 'B', 'b', 'C', 'a', 'c'}
setB = {'a', 'B', 'B', 'C', 'A', 'C'}

print(setA)
print(setB)
print(setA.difference_update(setB))

print(setA)
print(setB)
```

```python
# 6. set.discard(element)
setA = {'A', 'B', 'b', 'C', 'a', 'c'}
print(setA)

setA.discard('B')
print(setA)
```

```python
# 7. set.intersection(another_set) -
setA = {'A', 'B', 'b', 'C', 'a', 'c'}
setB = {'a', 'B', 'B', 'C', 'A', 'C'}

print(setA)
print(setB)

print(setA.intersection(setB))

print(setA)
print(setB)
```

```python
# 8. set.intersection_update(another_set) -
setA = {'A', 'B', 'b', 'C', 'a', 'c'}
setB = {'a', 'B', 'B', 'C', 'A', 'C'}
setC = {10,20,30,50,20,60}

print(setA)
print(setB)

print(setA.intersection_update(setB))

print(setA)
print(setB)

print(setB.intersection_update(setC))

print(setA)
print(setB)
```

```python
# 9. set.isdisjoint()
setA = {'A', 'B', 'b', 'C', 'a', 'c'}
setB = {'a', 'z', 'B', 'C', 'A', 'k'}
setC = {10,20,30,50,20,60, 'C'}


print(setA)
print(setB)
print(setC)
print(setA.isdisjoint(setB))
print(setB.isdisjoint(setA))
print(setA.isdisjoint(setC))
print(setC.isdisjoint(setB))
```

```python
# 10. set.issubset() -
A = {'A', 'B', 'b', 'C', 'a', 'c'}
B = {'A', 'c'}

print(A)
print(B)
print(A.issubset(B))
print(B.issubset(A))
```

```python
# 11. set.issuperset() -
A = {'A', 'B', 'b', 'C', 'a', 'c'}
B = {'A', 'c'}

print(A)
print(B)
print(A.issuperset(B))
print(B.issuperset(A))
```

```python
# 12. set.pop() -
A = {'A', 'B', 'b', 'C', 'a', 'c'}

print(A)
A.pop()
print(A)
A.pop()
print(A)
```

```python
# 13. set.remove(value) -
A = {'A', 'B', 'b', 'C', 'a', 'c'}

print(A)
A.remove('A')
print(A)
```
