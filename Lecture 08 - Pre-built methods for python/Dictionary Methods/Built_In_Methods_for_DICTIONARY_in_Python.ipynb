# **Pre-Built Functions (Built-In Functions)**


### **Built-In Functions in Dictionary**

```python
# 1. dict.clear()

empDictionary = {
    "empName": "Sagar",
    "empSalary": 250000,
    "empAge": 50,
    "empGender": "Male"
}

print(empDictionary)

# empDictionary.clear()

# print(empDictionary)
```

```python
# 2. dict.copy()

# newEmpDict = empDictionary

# print(newEmpDict)
# print(empDictionary)

# empDictionary.clear()

# print(newEmpDict)
# print(empDictionary)



newEmpDict = empDictionary.copy()
print(newEmpDict)
print(empDictionary)

empDictionary.clear()
print(newEmpDict)
print(empDictionary)
```

```python
# 3. dict.fromkeys()

productList = ["Watch", "Keyboard", "Android", "IOS", "Monitor"]
companyList = ["Fastrack", "Kreo", "Samsung", "Apple", "BenQ"]

prodDict = dict.fromkeys(companyList, productList)

print(prodDict)
```

```python
# 4. dict.get()

emp_name =  newEmpDict.get("empName")
emp_salary = newEmpDict.get("empSalary")

print(emp_name)
print(emp_salary)
```

```python
# 5. dict.keys()
newEmpKeys = newEmpDict.keys()
print(list(newEmpKeys))
```

```python
# 6. dict.value()
newEmpValues = newEmpDict.values()
print(tuple(newEmpValues))
```

```python
# 7. dict.items()
newEmpItems = newEmpDict.items()
print(list(newEmpItems))
```

```python
# 8. dict.pop()

newEmpDict.pop("empGender")
```
