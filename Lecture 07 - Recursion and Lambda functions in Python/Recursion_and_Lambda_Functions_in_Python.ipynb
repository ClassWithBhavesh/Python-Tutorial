# **Recursion in Python**

- it is a function which will call itself within the same function then it is a recursion and that kind of function is known as Recursive Function
- to stop or terminate the redudancy of recursive function we must have to use a **base case**

```python
import time as tl

start_time = tl.time()
def printNum(n):
  if n == 10:
    print(n)
  else:
    print(n)
    printNum(n + 1)
end_time = tl.time()

printNum(1)
print(end_time - start_time)
```

```python

startTime = tl.time()
for i in range(10):
  print(i)

endTime = tl.time()

printNum(1)
print(endTime - startTime)
```

# **Lambda Function - **

```python
result = lambda i: "Even" if i%2 == 0 else "Odd"

print(result(745))
```

```python
product = lambda x, y: x * y

print(product(20,12))
```
