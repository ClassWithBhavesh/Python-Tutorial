### **1. np.array()**

```python
import numpy as np

inventory = np.array(prodInSet)
print(inventory)
print(type(inventory))
```

### **2. np.arange()**

```python
rangeArr = np.arange(51, 101)
print(rangeArr)
print(type(rangeArr))
```

```python
rangeArr = np.arange(2, 101, 2)
print(rangeArr)
print(type(rangeArr))
```

```python
reverse5Table = np.arange(50, 4, -5)
print(reverse5Table)
```
